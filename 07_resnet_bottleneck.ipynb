# Task 7: ResNet-50 Style Bottleneck Block & Grouped Convolutions

A compact residual block using PyTorch `nn.Sequential`.

```python
import torch
import torch.nn as nn
```

### 1. ResNet-50 Bottleneck Block (1x1 down -> 3x3 conv -> 1x1 up + Skip)

```python
in_channels, bottleneck_dim, out_channels = 256, 64, 256

# Bottleneck Convolution Path
bottleneck_conv = nn.Sequential(
    nn.Conv2d(in_channels, bottleneck_dim, kernel_size=1),  # Reduce channels (256 -> 64)
    nn.ReLU(),
    nn.Conv2d(bottleneck_dim, bottleneck_dim, kernel_size=3, padding=1, groups=32), # Grouped Conv
    nn.ReLU(),
    nn.Conv2d(bottleneck_dim, out_channels, kernel_size=1)   # Expand channels (64 -> 256)
)
```

### 2. Forward Pass with Skip Connection

```python
x = torch.randn(1, 256, 14, 14)  # Input feature map

# Output = Conv Path + Original Input (Skip Connection!)
out = bottleneck_conv(x) + x

print('Input Shape :', x.shape)
print('Output Shape:', out.shape)
print('Parameter Count:', sum(p.numel() for p in bottleneck_conv.parameters()))
```
