# Task 8: Semantic Segmentation with Custom Dice Loss

Calculate pixel overlap loss (Soft Dice Loss) for image segmentation.

```python
import torch
import matplotlib.pyplot as plt
```

### 1. Define Soft Dice Loss Function in 3 Lines

```python
def dice_loss(pred_probs, target_mask):
    intersection = (pred_probs * target_mask).sum()
    union = pred_probs.sum() + target_mask.sum()
    return 1.0 - (2.0 * intersection + 1.0) / (union + 1.0)
```

### 2. Synthetic Target Mask & Prediction Test

```python
target = torch.zeros(1, 1, 10, 10)
target[0, 0, 3:7, 3:7] = 1.0  # Square mask target

# Simulated predicted probabilities
pred = target.clone() * 0.8  # Good prediction

loss_val = dice_loss(pred, target)
print(f'Soft Dice Loss Value: {loss_val.item():.4f}')
```

### 3. Display Target vs Predicted Mask

```python
fig, ax = plt.subplots(1, 2, figsize=(6, 3))
ax[0].imshow(target[0, 0], cmap='gray')
ax[0].set_title('Ground Truth Mask')
ax[1].imshow(pred[0, 0], cmap='gray')
ax[1].set_title('Predicted Mask')
plt.show()
```
