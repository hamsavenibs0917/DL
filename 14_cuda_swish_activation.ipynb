# Task 14: Swish Activation ($x \cdot \sigma(\beta x)$) & CUDA Kernel Concept

Super simple 1-line implementation of Swish activation and gradient computation.

```python
import torch
import matplotlib.pyplot as plt
```

### 1. Swish Activation Function & Derivative

```python
x = torch.linspace(-4, 4, 100, requires_grad=True)
beta = 1.0

# Swish formula: x * sigmoid(beta * x)
swish = x * torch.sigmoid(beta * x)

# Autograd backward pass for derivative f'(x)
swish.sum().backward()
grad_x = x.grad
```

### 2. Plot Swish Curve & Derivative

```python
plt.plot(x.detach().numpy(), swish.detach().numpy(), label="Swish")
plt.plot(x.detach().numpy(), grad_x.numpy(), label="Derivative", linestyle="--")

plt.title("Swish Activation and Gradient")
plt.xlabel("x")
plt.legend()
plt.grid()
plt.show()
```
