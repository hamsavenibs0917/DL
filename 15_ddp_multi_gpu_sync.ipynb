# Task 15: Distributed Data Parallel (DDP) Gradient Synchronization Concept

Understand how multiple worker nodes average gradients using All-Reduce.

```python
import torch
```

### 1. Simulate Gradients Computed on 2 Independent GPU Nodes

```python
node0_grad = torch.tensor([4, 2])  # Gradient computed on GPU Node 0
node1_grad = torch.tensor([8, 6])  # Gradient computed on GPU Node 1

print('Node 0 Gradient:', node0_grad)
print('Node 1 Gradient:', node1_grad)
```

### 2. All-Reduce Gradient Synchronization (Sum & Average)

```python
# DDP All-Reduce combines gradients across all nodes
synced_grad = (node0_grad + node1_grad) / 2.0

print('\nSynchronized Mean Gradient Across All Nodes:\n', synced_grad)
print('Now all nodes update their model weights using identical synced gradients!')
```
